```python
class Account:
    def __init__(self, prev_used, limit):
        """
        The Account class is used to track a user's data usage against a data limit.
        It stores:
        - the previously used amount
        - the maximum data limit

        The class also provides a method to verify whether adding new data usage
        will surpass the maximum data limit.
        """
        # Instance variables that store the user's previous data usage and data limit.
        self.prev_used = prev_used
        self.limit = limit
    
    def check_data_usage(self, data_used):
        """
        The function checks whether adding additional data usage exceeds the data limit.
        """
        # Calculate the total data usage after adding new usage.
        total_usage = self.prev_used + data_used
        if total_usage > self.limit:
            print("Data limit exceeded")
        else:
            print("Data usage is within the limit")
        # Update the stored previous usage with the new total.
        self.prev_used = total_usage

# The object Account is created where 9.4 has already been used and a maximum limit of 10.
# Numbers are taken from the previous function check_data_usage.

account1 = Account(prev_used=9.4, limit=10)

# Checks if adding 0.7 exceeds the limit.
account1.check_data_usage(0.7)
```
