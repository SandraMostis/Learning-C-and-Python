Heapsort builds a heap and repeatedly removes the smallest element (for descending order).
A minimum heap is used so that the smallest element is always at the root.

Start with list →[Mercury, Mars, Jupiter, Earth, Venus, Saturn]

First, build a minimum heap so that the smallest element is at the root.
The smallest element is Earth, so it is swapped with the last element and placed at the end.

Earth swapped with Saturn → [Mercury, Mars, Jupiter, Saturn, Venus, Earth]

The heap is then rebuilt (heapified) for the remaining elements.The new smallest element becomes Jupiter, which is at the root.

Jupiter is swapped with the last element of the remaining heap (Venus) and placed before Earth →[Mercury, Mars, Venus, Saturn, Jupiter, Earth]

The heap is rebuilt again. The next smallest element is Mars, which is swapped with the last element of the remaining heap and placed next.

This process continues:

Mercury is placed next (after heapify and swap) then Saturn and finally Venus

Each time, the root (smallest element) is swapped with the last element of the unsorted portion and the heap is rebuilt.

Final sorted list (descending alphabetical order):
[Venus, Saturn, Mercury, Mars, Jupiter, Earth]

```python
from collections import defaultdict

class Heap:
    """
    A max-heap implementation using a list.
    The largest element is always at the root (index 0).
    """

    def __init__(self, data=None):
        """
        Initialize the heap.
        data: Optional list of initial values.
        """
        if data is None:
            self.data = []
        else:
            self.data = data

    def is_empty(self):
        """
        Checks if the heap is empty.
        Returns True if heap is empty, otherwise False.
        """
        return len(self.data) == 0

    def parent(self, index):
        """
        Returns the index of the parent node. Index is the index of current node.
        """
        return (index - 1) // 2

    def left(self, index):
        """
        Returns the index of the left child. Index is the index of current node.
        """
        return index * 2 + 1

    def right(self, index):
        """
        Returns the index of the right child. Index is the index of current node.
        """
        return index * 2 + 2

    def insert(self, value):
        """
        Inserts a value into the heap and maintain heap property.
        """
        pos = len(self.data)
        self.data.append(value)

        # Move the value up until heap property is restored
        while pos > 0 and self.data[self.parent(pos)] < value:
            self.data[pos] = self.data[self.parent(pos)]
            pos = self.parent(pos)

        self.data[pos] = value

    def extract(self):
        """
        Removes and return the maximum element from the heap.
        """
        max_value = self.data[0]

        # Replace root with last element and shrink heap
        self.data[0] = self.data[len(self.data) - 1]
        self.data.pop()

        # Restore heap property
        self.move_down(0, len(self.data) - 1)

        return max_value

    def move_down(self, first, last):
        """
        Restore heap property by moving a node down the tree.
        -first: Index of the node to move down.
        -last: Last valid index in the heap.
        """
        while self.left(first) <= last:
            larger = self.left(first)

            # Select the larger of the two children
            if (self.right(first) <= last and
                self.data[self.left(first)] < self.data[self.right(first)]):
                larger = self.right(first)

            # If heap property is satisfied, stop
            if self.data[first] >= self.data[larger]:
                break

            # Swap with larger child
            self.data[first], self.data[larger] = self.data[larger], self.data[first]
            first = larger


def heap_sort(data):
        """
        Perform heap sort on a list in descending alphabetical order.
        data: List of elements to sort.
        """
        heap = Heap(data)

        # Build the max-heap
        for i in range(len(heap.data)//2 - 1, -1, -1):
            heap.move_down(i, len(heap.data) - 1)

        # Extract elements one by one and rebuild heap
        for i in range(len(heap.data) - 1, 0, -1):
            # Move current root (largest) to the end
            heap.data[0], heap.data[i] = heap.data[i], heap.data[0]

            # Restore heap property for remaining elements
            heap.move_down(0, i - 1)

        return heap.data    


# Please note that I have found this task challenging and I have used Gen AI to complete the assignment!

class SolarSystem:
    """
    A class that stores planet names using a max-heap.
    The heap maintains the property that the largest (alphabetically)
    element is always at the root, which supports descending order.
    """

    def __init__(self, data):
        """
        Initialize the SolarSystem with a list of planet names.
        The input list is converted into a heap structure so that
        heap operations can be applied efficiently.
        data: List of planet names (strings).
        """
        # Create a Heap object using the provided list
        # This stores the data internally but does not yet guarantee heap order
        self.heap = Heap(data)

        # Build the max-heap by applying move_down from the last non-leaf node down to the root
        # This ensures the heap property is satisfied.
        for i in range(len(self.heap.data)//2 - 1, -1, -1):
            # move_down ensures the subtree rooted at index i satisfies heap rules
            self.heap.move_down(i, len(self.heap.data) - 1)

    def __str__(self):
        """
        Return a string representation of the heap.
        This shows the internal array representation of the heap,
        not a fully sorted list.
        """
        # Convert the heap data list to string for easy printing
        return str(self.heap.data)


# Example usage
planets = ["Mercury", "Mars", "Jupiter", "Earth", "Venus", "Saturn"]

# Create SolarSystem object
solar_system = SolarSystem(planets)

print("Heap representation:")
print(solar_system)

print("\nSorted (descending alphabetical):")
print(heap_sort(planets.copy()))
```
