Depth-first search (DFS) explores as far as possible along each branch before backtracking. The traversal does not consider edge weights, only adjacency and visit status.

Step-by-step traversal:

Start at Vertex 5 → mark as visited
Move to adjacent unvisited vertex 6 (only unvisited neighbour of 5)

Visit Vertex 6 → mark as visited
Vertex 5 already visited → move to 4 (next unvisited adjacent vertex)

Visit Vertex 4 → mark as visited
Vertex 5 already visited → move to 1 (next unvisited adjacent vertex)

Visit Vertex 1 → mark as visited
Visit next unvisited adjacent vertex 2 (chosen before 3)

Visit Vertex 2 → mark as visited
Adjacent vertex 4 already visited → backtrack (no unvisited neighbours)

Backtrack to Vertex 1, visit next unvisited vertex 3

Visit Vertex 3 → mark as visited
Adjacent vertex 4 already visited → backtrack (no unvisited neighbours remain)

There are no more unvisited vertices reachable, so the traversal stops.

Visiting order: 5 → 6 → 4 → 1 → 2 → 3

```python
from collections import defaultdict

class DirectedWeightGraph:
    """A directed graph where edges have weights.
       The graph is represented using:
       - A set for vertices
       - An adjacency list (dictionary) for edges"""

    def __init__(self):
        """Initialize an empty graph.
           vertices: stores all vertices in the graph
           edges: maps each vertex to a list of (neighbour, weight) pairs
        """
        self.vertices = set()
        self.edges = defaultdict(list)

    def add_vertex(self, v):
        """Add a vertex to the graph."""
        self.vertices.add(v)

    def add_edge(self, v, u, weight):
        """
        Add a directed edge from vertex v to vertex u with a given weight.
        v: Start vertex
        u: End vertex
        weight: Weight of the edge
        """
        self.edges[v].append((u, weight))

    def __delitem__(self, v):
        """Remove a vertex and all associated edges."""
        # Remove the vertex from the set of vertices
        if v in self.vertices:
            self.vertices.remove(v)
        
        # Remove all outgoing edges from v
        if v in self.edges:
            del self.edges[v]
        
        # Remove all incoming edges pointing to v
        for vertex in self.edges:
            self.edges[vertex] = [(neighbour, w) for (neighbour, w) in self.edges[vertex] if neighbour != v]

    def __str__(self):
        """Return a readable string representation of the graph."""
        result = ""

        # Sort vertices for consistent output
        for v in sorted(self.vertices):
            result += f"{v} -> "
            # lift of outgoing edges
            if v in self.edges:
                result += ", ".join([f"({u}, w={w})" for (u, w) in self.edges[v]])
            result += "\n"
        return result
            
graph = DirectedWeightGraph()

# Add vertices
for v in [1, 2, 3, 4, 5, 6]:
    graph.add_vertex(v)

# Add edges (directed with weights)
graph.add_edge(1, 2, 2)
graph.add_edge(2, 4, 4)
graph.add_edge(1, 4, 6)
graph.add_edge(1, 3, 5)
graph.add_edge(3, 4, 2)
graph.add_edge(4, 5, 6)
graph.add_edge(6, 4, 4)
graph.add_edge(5, 6, 3)
graph.add_edge(6, 5, 3)
print(graph)
```
