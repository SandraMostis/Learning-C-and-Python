```python
class LibraryItem:

    """The LibraryItem class represents a general item in a library system.

    It stores common information shared by all library items, including:
    - item ID
    - title
    - publication date

    The class also provides methods to create and display a reference
    for a library item. It acts as a parent class for more specific
    library item types such as books and magazines."""

    def __init__(self, item_id=None, title=None, publication_date=None):

        # Protected attribute storing the library item's unique ID
        self._item_id = item_id

        # Protected attribute storing the title of the item
        self._title = title

        # Protected attribute storing the publication date of the item
        self._publication_date = publication_date
    
    def create(self, l_id: int, t: str, d: str):
        """
        Update the library item's details.

        Parameters:
        l_id (int): The item ID.
        t (str): The title of the item.
        d (str): The publication date.
        """

        # Assign a new ID to the item
        self._item_id = l_id

        # Assign a new title to the item
        self._title = t

        # Assign a new publication date
        self._publication_date = d

    def print_reference(self):
        """
        Display a formatted reference for the library item.
        """

        # Print the item's reference using title, publication date and ID
        print(f"{self._title} ({self._publication_date}) - ID: {self._item_id}")


class Book(LibraryItem):
    """
    The Book class represents a book in the library system.

    It inherits from the LibraryItem class and adds:
    - author information

    The class overrides the print_reference method to include
    the author's name in the reference output.
    """

    def __init__(self):

        # Call the constructor of the parent class to initialise inherited attributes
   
        super().__init__()

        # Private attribute storing the author's name
        self.__author = None

    def create(self, l_id: int, t: str, d: str, a: str):
        """
        Create or update a book's details.

        Parameters:
        l_id (int): The item ID.
        t (str): The book title.
        d (str): The publication date.
        a (str): The author's name.
        """

        # Use the parent class method to set
        # shared library item information
        super().create(l_id, t, d)
        # Store the author's name in the private attribute
        self.__author = a

    def print_reference(self):

        """Display a formatted reference for the book."""
        # Print the author's name followed by
        # the inherited title, publication date and ID
        print(f"{self.__author}, {self._title} ({self._publication_date}) - ID: {self._item_id}")

class Magazine(LibraryItem):
    """
    The Magazine class represents a magazine in the library system.

    It inherits from the LibraryItem class and adds:
    - issue number information

    The class overrides the print_reference method to include
    the magazine issue number in the reference output.
    """
    def __init__(self):

        # Call the constructor of the parent class
        # to initialise inherited attributes
        super().__init__()
        # Private attribute storing the issue number
        self.__issue_number = None

    def create(self, l_id: int, t: str, d: str, i: int):
        """
        Create or update a magazine's details.
        Parameters:
        l_id (int): The item ID.
        t (str): The magazine title.
        d (str): The publication date.
        i (int): The issue number.
        """
        # Use the parent class method to initialise
        # common library item details
        super().create(l_id, t, d)
        # Store the issue number in the private attribute
        self.__issue_number = i

    def print_reference(self):
        """
        Display a formatted reference for the magazine.
        """
        # Print the magazine title, issue number, publication date and item ID
        print(f"{self._title}, Issue {self.__issue_number} ({self._publication_date}) - ID: {self._item_id}")


# Create a Book object
book1 = Book()
# Add values to the book object
book1.create(100, "The Hitchhiker's Guide to the Galaxy", "1979", "Douglas Adams")
# Display the formatted book reference
book1.print_reference()
# Create a Magazine object
magazine1 = Magazine()
# Add values to the magazine object
magazine1.create(200, "Hello!", "2026", 1941)
# Display the formatted magazine reference
magazine1.print_reference()
```
