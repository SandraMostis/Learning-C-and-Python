## Exercise 4 i)

We start breadth-first traversal by using a queue and visiting nodes level by level from left to right.

First, we add the root node 40 to the queue.

Queue: [40] 
We visit 40 and remove it from the queue. Then we add its children (left to right) to the queue: 50 and 20.

Queue: [50, 20]
We visit 50 next and remove it from the queue. We then add the children of 50: 10 and 30 to the queue.

Queue: [20, 10, 30]
Next, we visit 20 and remove it from the queue. Node 20 has one child to the left but we have 60 to the right so we add it to the queue:

Queue: [10, 30, 60]

Now we continue visiting in order:

Visit 10 → it has no children and we dequeue 10.

Queue: [30, 60]

Visit 30 → it has no children, we dequeue 30.

Queue: [60]

Visit 60 → it has no children, we dequeue 60.

Queue: []

The queue is now empty, so we stop.

Visiting order: 40 → 50 → 20 → 10 → 30 → 60

## Exercise 4 ii)

This is an AVL tree.

An AVL tree is a binary search tree where, for every node, the difference in height between the left and right subtrees is ≤ 1.

For node 40, the left subtree has height 2 and the right subtree has height 2. The difference is 0, therefore it is balanced.

For node 50, the left subtree (node 10) has height 1 and the right subtree (node 30) has height 1. The difference is 0, therefore it is balanced.

For node 20, the left subtree has height 0 and the right subtree has height 1. The difference is 1, therefore it is balanced.

Leaf nodes are automatically balanced. 

Since all nodes have a height difference of at most 1, the tree satisfies the AVL balancing condition.



```python
class Node:
    """ Represents a binary tree node"""

    def __init__(self, val, l=None, r=None):

        """Initialises a tree node and takes 3 arguments:
        val is the value stored in the node
        l is the value for the left child node
        r is the value for the right child node"""
        
        # Store the value of the current node
        self.value = val 
        # Left child is assigned
        self.left = l 
        # Right child is assigned
        self.right = r
        
    def __str__(self):

        """Returns a string representation of the tree.
        Leaf nodes return only their value and non-leaf nodes include their left and right subtrees."""

        # If the node has no children it returns its own value

        if self.left == None and self.right == None:
            return str(self.value)
        
        # Else it builds the subtree recursively 
        else:
            return str(self.value) + "({})({})".format(
                str(self.left) if self.left != None else "", 
                str(self.right) if self.right != None else "")

# Creates a binary tree from the diagram at point i)
my_tree = Node("40", 
               Node("50", 
                    Node("10"), 
                    Node("30")),
               Node("20",
                    None,
                    Node("60")))
print(my_tree)
```

```python
# Node class
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

# Function to check if value is an integer
def is_int(value):
    return isinstance(value, int)

# Function to check if value is a string
def is_str(value):
    return isinstance(value, str)


class Search_tree:
    """Represent a search tree that separates integers and strings."""

    def __init__(self):
        """Create an empty search tree."""
        self.root = None

    def insert(self, val, node="root"):
        """
        Insert a value into the tree.

        Integers are inserted into the left subtree of the root.
        Strings are inserted into the right subtree of the root.
        """

        # Start insertion at the root, first value becomes the root
        if node == "root":

            # Create root if tree is empty
            if self.root is None:
                self.root = Node(val)
                return self.root

            # Integers go into left subtree
            if is_int(val):
                self.root.left = self.insert(val, self.root.left)

            # Strings go into right subtree
            elif is_str(val):
                self.root.right = self.insert(val, self.root.right)

            # Raise value error if the insert is not int or str
            else:
                raise ValueError("Only integers and strings are allowed")

            return self.root

        # Create a new node if subtree is empty
        if node is None:
            return Node(val)

        if val < node.value:
            node.left = self.insert(val, node.left)

        elif val > node.value:
            node.right = self.insert(val, node.right)

        return node


tree = Search_tree()

tree.insert(10)          # Root
tree.insert(5)           # Integer -> left of root
tree.insert(20)          # Integer -> left subtree
tree.insert("apple")     # String -> right of root
tree.insert("banana")    # String -> right subtree

print(tree.root.value)
print(tree.root.left.value)
print(tree.root.left.right.value)
print(tree.root.right.value)
print(tree.root.right.right.value)
```
